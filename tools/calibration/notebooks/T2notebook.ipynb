```python
import numpy as np, matplotlib.pyplot as plt, matplotlib
matplotlib.use('TkAgg')
%matplotlib inline
from T2 import socket, cluster_heimdall as ch
import time
```

```python
fl = "/operations/T2/cluster_output/T1_output1732984528.csv"
tab = ch.parse_candsfile(fl)
print(len(tab))
```

```python
def plot_cands(tab,t2=None):
    
    plt.figure(figsize=(13,15))
    
    plt.subplot(3,1,1)
    
    x = np.asarray(tab["ibeam"])
    y = np.asarray(tab["dm"])
    sel = np.asarray(tab["ibox"])
    
    plt.plot(x[sel<5],y[sel<5],'k.')
    plt.plot(x[sel==5],y[sel==5],'r+')
    plt.plot(x[sel==6],y[sel==6],'gx')
    
    if t2 is not None:
        x = np.asarray(t2["ibeam"])
        y = np.asarray(t2["dm"])
        sel = np.asarray(t2["ibox"])
    
        plt.plot(x[sel<5],y[sel<5],'ko',markersize=15)
        plt.plot(x[sel==5],y[sel==5],'ro',markersize=15)
        plt.plot(x[sel==6],y[sel==6],'go',markersize=15)
        
    
    plt.xlabel("Beam")
    plt.ylabel("DM")
    
    
    plt.subplot(3,1,2)
    
    x = np.asarray(tab["mjds"])
    x -= np.min(x)
    x *= 86400.
    y = np.asarray(tab["snr"])
    sel = np.asarray(tab["ibox"])
    
    plt.plot(x[sel<5],y[sel<5],'k.')
    plt.plot(x[sel==5],y[sel==5],'r+')
    plt.plot(x[sel==6],y[sel==6],'gx')
    
    if t2 is not None:
        x = np.asarray(t2["mjds"])
        x -= np.min(x)
        x *= 86400.
        y = np.asarray(t2["snr"])
        sel = np.asarray(t2["ibox"])
    
        plt.plot(x[sel<5],y[sel<5],'ko',markersize=15)
        plt.plot(x[sel==5],y[sel==5],'ro',markersize=15)
        plt.plot(x[sel==6],y[sel==6],'go',markersize=15)
    
    plt.xlabel("Time (s)")
    plt.ylabel("SNR")
    
    plt.subplot(3,1,3)
    
    
    bms = np.unique(np.asarray(tab["ibeam"]))
    stds = np.zeros(len(bms))
    for i in np.arange(len(bms)):
        tt = tab[tab["ibeam"]==bms[i]]["dm"]
        stds[i] = np.std(tt)
        stds[i] *= len(tt)
        
    plt.plot(bms,stds,'ko')
    
    print(len(tab2))
    
    plt.show()
    
```

```python
def flag_tab(tab,stat=8e5):
    
    t0 = time.time()
    tab2 = tab
    bms = np.unique(np.asarray(tab["ibeam"]))
    stds = np.zeros(len(bms))
    for i in np.arange(len(bms)):
        tt = tab[tab["ibeam"]==bms[i]]["dm"]
        stds[i] = np.std(tt)
        stds[i] *= len(tt)
        if stds[i]>stat:
            print(f"Flagging beam {bms[i]}")
            tab2 = tab2[tab2["ibeam"]!=bms[i]]
    t1 = time.time()
    print(f"Flagging took {t1-t0} s")
        
    return tab2
    
    
```

```python
tab2 = flag_tab(tab)
```

```python
plot_cands(tab2)
```

```python
t2 = ch.get_peak(tab2)
```

```python
plot_cands(tab2,t2=t2)
```
