```python
import numpy as np, matplotlib.pyplot as plt
import yaml
from yaml import Loader
```

```python
with open("beamformer_weights.yaml.bak3") as f:
    ow = yaml.load(f,Loader)
f.close()
with open("bfweights.yaml") as f:
    nw = yaml.load(f,Loader)
f.close()
with open("bfweights2.yaml") as f:
    nw2 = yaml.load(f,Loader)
f.close()
```

```python
ao = np.array(nw['cal_solutions']['antenna_order'])
nw_d = np.array(nw['cal_solutions']['delays'])
nw2_d = np.array(nw2['cal_solutions']['delays'])
ow_d = np.array(ow['cal_solutions']['delays'])
```

```python
plt.figure(figsize=(12,10))
plt.plot(nw_d[:,0]-ow_d[:,0],'o')
plt.plot(nw2_d[:,0]-ow_d[:,0],'o')
plt.ylim(-20,20)
```

```python
correct_ao = ow['cal_solutions']['antenna_order']
correct_ao[27] = 101
correct_ao[28] = 9
correct_ao[29] = 8
```

```python
for i in (list(np.arange(68,100))):
    correct_ao.append(i)
```

```python
ao = list(ao)
```

```python
correct_delays = []
for i in correct_ao:
    
    oi = 0
    for j,v in enumerate(ao):
        if v==i:
            oi=j
    print(oi,i,nw2['cal_solutions']['delays'][oi])
    correct_delays.append(nw2['cal_solutions']['delays'][oi])

    
```

```python
nw['cal_solutions']['delays'] = correct_delays
```

```python
with open("bfweights2.yaml","w") as f:
    yaml.dump(nw2,stream=f,Dumper=yaml.Dumper)
f.close()
```
