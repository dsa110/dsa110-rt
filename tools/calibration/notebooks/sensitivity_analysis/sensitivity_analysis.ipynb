```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
d = np.load('sensitivities.npz')
times = d['times']
noise_5_50 = d['noise_5_50']
noise_5_100 = d['noise_5_100']
noise_5_200 = d['noise_5_200']
noise_5_400 = d['noise_5_400']
noise_300_1000 = d['noise_300_1000']
times -= 7./24.
times -= np.floor(times[0])
```

```python
plt.figure(figsize=(12,6))

plt.semilogy(times,noise_5_50,label='noise_5_50')
plt.semilogy(times,noise_5_100,label='noise_5_100')
plt.semilogy(times,noise_5_200,label='noise_5_200')
plt.semilogy(times,noise_5_400,label='noise_5_400')
plt.semilogy(times,noise_300_1000,label='noise_300_1000')
plt.xlabel('Time of day',fontsize=14)
plt.ylabel('SEFD-ish',fontsize=14)
plt.legend(fontsize=14)
plt.ylim(8e3,15e3)
plt.grid()
```

```python
for i in np.arange(len(times)):
    print(times[i],d['fls'][i])
```

```python
import sens_funcs as sf

fl = '/operations/correlator/2022-07-01T07:00:23_sb05.hdf5'
calfile = 'beamformer_weights_sb05_1411+522_2022-06-28T03:40:24.dat'
uvdist,ant1,ant2,pl_A,pl_B = sf.calc_noise(fl,calfile)

# make uv plot of noise (in Jy) v baseline length
def make_uvplot(uvdist,pl_A,pl_B):

    plt.figure(figsize=(12,7))
    plt.loglog(uvdist,pl_B,'.',color='mediumslateblue')
    plt.loglog(uvdist,pl_A,'.',color='lightcoral')
    
    plt.xlim(5.,2500.)
    plt.xlabel('UV distance (m)',fontsize=14)
    plt.ylabel('SEFD (Jy)',fontsize=14)
    plt.grid()
    plt.show()
        
make_uvplot(uvdist,pl_A,pl_B)
```
